# Creating the dataset

```python
from PhaseEstimation.inputs import Diag, Mps, Brick
import numpy as np
import tqdm.notebook as tqdm
from pathlib import Path
import pickle
```

```python
num_spins = 8
side = 3
path = "./data/inputs/"
Path(path).mkdir(parents=True, exist_ok=True)

ks = np.linspace(0, 1, side) 
hs = np.linspace(0, 2, side) 
```

### Diagonalization of the Hamiltonian Matrix

```python
progress = tqdm.tqdm(range(len(ks) * len(hs)))
for k in ks:
    for h in hs:
        Diag.make(k, h, num_spins, path, label="diag")
        progress.update(1)
```

### DMRG

```python
chi = 8
progress = tqdm.tqdm(range(len(ks) * len(hs)))
for k in ks:
    for h in hs:
        Mps.make(k, h, num_spins, chi, path, label="dmrg")
        progress.update(1)
```

### MPS (DMRG) $\to$ Brick-Layered Ansatz

```python
depth = 4
for k in ks:
    for h in hs:
        filename = Path(path) / f"dmrg_k{k}_h{h}.pkl"
        with open(filename, "rb") as f:
            mps = pickle.load(f)
        Brick.make(k, h, mps, depth, path, label="brick")
```
